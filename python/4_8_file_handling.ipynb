**1. Opening & Reading a file**

```python
# open a file in read mode
file = open("notes.txt", "r")

# read the content
content = file.read()

print(content)

file.close()
```

**2. Using with to read (auto-closing)**

```python
with open("notes.txt", "r") as file:
    content = file.read()
    print(content)
```

**3. Writing to a File**

```python
with open("output.txt", "w") as file:
    file.write("This is my first file")
```

```python
with open("output.txt", "w") as file:
    file.write("This is my second file.")
```

```python
with open("output.txt", "a") as file:
    file.write("\nThis line was added later.")
```

```python
with open("output.txt", "a") as file:
    file.write("\nThis is the third line.")
```

```python
with open("output.txt", "r") as file:
    for line in file:
        print(line)
```
